```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Step 1: Loading the Dataset

```python
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Crime.csv")


print(f"Dataset Shape: {df.shape[0]} rows x {df.shape[1]} columns\n")
df.head()
print("First 5 rows:")
```

## Step 2: Basic Statistical Analysis

```python
print("Column Info:")
print(df.info())
print("\n" + "="*70 + "\n")

print("Summary Statistics (Numerical Columns):")
df.describe()
```

```python
print("Skewness of numerical features:")
print(df.select_dtypes(include=[np.number]).skew().round(4))
print("\nKurtosis of numerical features:")
print(df.select_dtypes(include=[np.number]).kurtosis().round(4))
print("\nVariance of numerical features:")
print(df.select_dtypes(include=[np.number]).var().round(4))
```

## Step 3: Handling Missing Data

```python
print("Missing Values per Column:")
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    "Missing Count": missing,
    "Percentage (%)": missing_pct.round(3)
})
print(missing_df[missing_df["Missing Count"] > 0] if missing_df["Missing Count"].sum() > 0 else "No missing values found!")

if missing.sum() > 0:
    plt.figure(figsize=(12, 6))
    sns.heatmap(df.isnull(), cbar=True, yticklabels=False, cmap="viridis")
    plt.title("Missing Data Heatmap", fontsize=14)
    plt.tight_layout()
    plt.show()
```

```python
print("Handling missing values...")
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if df[col].dtype in [np.float64, np.int64]:
            df[col].fillna(df[col].median(), inplace=True)
            print(f"  Filled numeric column '{col}' with median: {df[col].median():.4f}")
        else:
            df[col].fillna(df[col].mode()[0] if not df[col].mode().empty else "Unknown", inplace=True)
            print(f"  Filled categorical column '{col}' with mode")

print(f"\nMissing values after imputation: {df.isnull().sum().sum()}")
```

## Step 4: Data Cleaning

```python
print("Data Cleaning in progress...\n")

duplicates = df.duplicated().sum()
print(f"Duplicate rows: {duplicates}")
if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print(f"  Removed {duplicates} duplicate rows")

if (df["crmrte"] < 0).any():
    print(f"  Found {(df['crmrte'] < 0).sum()} negative crime rates - setting to NaN and imputing")
    df.loc[df["crmrte"] < 0, "crmrte"] = np.nan
    df["crmrte"].fillna(df["crmrte"].median(), inplace=True)

cat_cols = df.select_dtypes(include=["object"]).columns
for col in cat_cols:
    unique_vals = df[col].nunique()
    print(f"  Column '{col}': {unique_vals} unique values")

df.reset_index(drop=True, inplace=True)

print(f"\nCleaned dataset: {df.shape[0]} rows x {df.shape[1]} columns")
```

## Step 5: Data Transformation

```python
print("Data Transformation in progress...\n")

le = LabelEncoder()
cat_cols = df.select_dtypes(include=["object"]).columns.tolist()
print(f"Categorical columns to encode: {cat_cols}")

for col in cat_cols:
    df[col + "_encoded"] = le.fit_transform(df[col])
    print(f"  '{col}' encoded: {dict(zip(le.classes_, range(len(le.classes_))))}")

df_transformed = df.copy()

scaler = StandardScaler()
numeric_cols_to_scale = ["crmrte", "density", "taxpc", "polpc", "avgsen", "pctmin", "pctymle"]
existing_cols = [c for c in numeric_cols_to_scale if c in df_transformed.columns]
df_transformed[[c + "_scaled" for c in existing_cols]] = scaler.fit_transform(df_transformed[existing_cols])

print(f"\nScaled columns: {existing_cols}")
print("\nPreview of transformed data:")
df_transformed[[c + "_scaled" for c in existing_cols]].head()
```

## Step 6: Univariate Analysis (3 Visualizations)

### Visualization 1: Histogram of Crime Rate (`crmrte`)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df["crmrte"], bins=30, edgecolor="black", color="steelblue", alpha=0.7, density=True)
df["crmrte"].plot.kde(ax=axes[0], color="darkred", linewidth=2)
axes[0].set_title("Distribution of Crime Rate (crmrte)", fontsize=13)
axes[0].set_xlabel("Crime Rate")
axes[0].set_ylabel("Density")
axes[0].axvline(df["crmrte"].mean(), color="red", linestyle="--", label=f"Mean: {df['crmrte'].mean():.4f}")
axes[0].axvline(df["crmrte"].median(), color="green", linestyle="--", label=f"Median: {df['crmrte'].median():.4f}")
axes[0].legend()

axes[1].boxplot(df["crmrte"].dropna(), vert=True, patch_artist=True,
                boxprops=dict(facecolor="steelblue", alpha=0.7))
axes[1].set_title("Boxplot of Crime Rate", fontsize=13)
axes[1].set_ylabel("Crime Rate")
axes[1].set_xticklabels(["crmrte"])

plt.tight_layout()
plt.show()

print(f"Crime Rate Statistics:")
print(f"   Mean: {df['crmrte'].mean():.6f}  |  Median: {df['crmrte'].median():.6f}")
print(f"   Std:  {df['crmrte'].std():.6f}  |  Skew: {df['crmrte'].skew():.4f}")
```

### Visualization 2: Count Plot of Region

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

colors = sns.color_palette("Set2", n_colors=df["region"].nunique())
sns.countplot(data=df, x="region", hue="region", palette=colors, legend=False, ax=axes[0], order=df["region"].value_counts().index)
axes[0].set_title("Count of Counties by Region", fontsize=13)
axes[0].set_xlabel("Region")
axes[0].set_ylabel("Number of Counties")
for p in axes[0].patches:
    axes[0].annotate(f"{int(p.get_height())}", (p.get_x() + p.get_width()/2., p.get_height()),
                     ha="center", va="bottom", fontsize=10)

region_counts = df["region"].value_counts()
axes[1].pie(region_counts.values, labels=region_counts.index, autopct="%1.1f%%",
            colors=sns.color_palette("Set2", n_colors=len(region_counts)),
            explode=[0.02]*len(region_counts), shadow=True)
axes[1].set_title("Region Distribution (Pie Chart)", fontsize=13)

plt.tight_layout()
plt.show()
```

### Visualization 3: Violin Plot of Population Density

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.violinplot(y=df["density"], color="lightblue", ax=axes[0, 0])
axes[0, 0].set_title("Violin Plot of Population Density", fontsize=13)
axes[0, 0].set_ylabel("Density")

sns.kdeplot(data=df, x="density", fill=True, color="teal", alpha=0.4, ax=axes[0, 1])
sns.kdeplot(data=df, x="density", color="darkblue", linewidth=2, ax=axes[0, 1])
axes[0, 1].set_title("KDE Plot of Population Density", fontsize=13)
axes[0, 1].set_xlabel("Density")

sns.histplot(data=df, x="taxpc", bins=25, kde=True, color="coral", ax=axes[1, 0])
axes[1, 0].set_title("Distribution of Tax Per Capita", fontsize=13)
axes[1, 0].set_xlabel("Tax Per Capita")

sns.countplot(data=df, x="smsa", hue="smsa", palette="pastel", legend=False, ax=axes[1, 1])
axes[1, 1].set_title("Count of SMSA (Metropolitan Area)", fontsize=13)
for p in axes[1, 1].patches:
    axes[1, 1].annotate(f"{int(p.get_height())}", (p.get_x() + p.get_width()/2., p.get_height()),
                         ha="center", va="bottom", fontsize=10)

plt.tight_layout()
plt.show()
```

## Step 7: Bivariate Analysis (3 Visualizations)

### Visualization 1: Scatter Plot — Crime Rate vs Police Per Capita

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.regplot(data=df, x="polpc", y="crmrte", ax=axes[0],
            scatter_kws={"alpha": 0.5, "color": "steelblue"},
            line_kws={"color": "red", "linewidth": 2})
axes[0].set_title("Crime Rate vs Police Per Capita (with regression)", fontsize=13)
axes[0].set_xlabel("Police Per Capita")
axes[0].set_ylabel("Crime Rate")

sns.scatterplot(data=df, x="polpc", y="crmrte", hue="region",
                palette="Set2", alpha=0.7, ax=axes[1])
axes[1].set_title("Crime Rate vs Police Per Capita (by Region)", fontsize=13)
axes[1].set_xlabel("Police Per Capita")
axes[1].set_ylabel("Crime Rate")
axes[1].legend(bbox_to_anchor=(1.05, 1), loc="upper left")

corr_val = df["crmrte"].corr(df["polpc"])
print(f"Correlation (crmrte vs polpc): {corr_val:.4f}")
plt.tight_layout()
plt.show()
```

### Visualization 2: Correlation Heatmap of Numerical Features

```python
corr_cols = ["crmrte", "prbarr", "prbconv", "prbpris", "avgsen", "polpc",
             "density", "taxpc", "pctmin", "pctymle",
             "wcon", "wtuc", "wtrd", "wfir", "wser", "wmfg", "wfed", "wsta", "wloc", "mix"]
existing_corr_cols = [c for c in corr_cols if c in df.columns]

corr_matrix = df[existing_corr_cols].corr()

plt.figure(figsize=(16, 12))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="RdBu_r", center=0,
            mask=mask, square=True, linewidths=0.5,
            cbar_kws={"shrink": 0.8})
plt.title("Correlation Heatmap of Crime Dataset Features", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()

print("Top Correlations with Crime Rate (crmrte):")
crmrte_corr = corr_matrix["crmrte"].drop("crmrte").sort_values(ascending=False)
print(crmrte_corr.to_string())
```

### Visualization 3: Boxplot — Crime Rate by Region

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x="region", y="crmrte", palette="Set2", ax=axes[0])
axes[0].set_title("Crime Rate Distribution by Region", fontsize=13)
axes[0].set_xlabel("Region")
axes[0].set_ylabel("Crime Rate")

sns.boxplot(data=df, x="smsa", y="crmrte", palette="Set3", ax=axes[1])
axes[1].set_title("Crime Rate Distribution by SMSA (Metropolitan Status)", fontsize=13)
axes[1].set_xlabel("SMSA")
axes[1].set_ylabel("Crime Rate")

plt.tight_layout()
plt.show()

print("Mean Crime Rate by Region:")
print(df.groupby("region")["crmrte"].mean().round(6).to_string())
print("\nMean Crime Rate by SMSA:")
print(df.groupby("smsa")["crmrte"].mean().round(6).to_string())
```

## Step 8: Multivariate Analysis (3 Visualizations)

### Visualization 1: Pairplot of Key Features by Region

```python
pair_cols = ["crmrte", "polpc", "density", "taxpc", "pctmin", "avgsen"]
existing_pair_cols = [c for c in pair_cols if c in df.columns]

g = sns.pairplot(df[existing_pair_cols + ["region"]], hue="region",
                 palette="Set2", diag_kind="kde",
                 plot_kws={"alpha": 0.6, "s": 40},
                 corner=True)
g.fig.suptitle("Pairplot of Key Crime Features by Region", y=1.02, fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```

### Visualization 2: Clustered Heatmap — Crime Rate by Region & Year

```python
pivot_table = df.pivot_table(values="crmrte", index="region", columns="year", aggfunc="mean")

plt.figure(figsize=(10, 6))
sns.heatmap(pivot_table, annot=True, fmt=".4f", cmap="YlOrRd",
            linewidths=0.5, cbar_kws={"label": "Mean Crime Rate"})
plt.title("Average Crime Rate by Region and Year", fontsize=14, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Region")
plt.tight_layout()
plt.show()

plt.figure(figsize=(16, 12))
sns.clustermap(corr_matrix, annot=True, fmt=".2f", cmap="RdBu_r", center=0,
               linewidths=0.5, figsize=(14, 12))
plt.suptitle("Clustered Correlation Heatmap", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

### Visualization 3: Parallel Coordinates / Radar — Wage Components by Region

```python
wage_cols = ["wcon", "wtuc", "wtrd", "wfir", "wser", "wmfg", "wfed", "wsta", "wloc"]
existing_wage_cols = [c for c in wage_cols if c in df.columns]

wage_by_region = df.groupby("region")[existing_wage_cols].mean()

from math import pi

categories = existing_wage_cols
N = len(categories)
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(polar=True))
colors = sns.color_palette("Set2", n_colors=len(wage_by_region))

for idx, (region_name, row) in enumerate(wage_by_region.iterrows()):
    values = row.values.tolist()
    values += values[:1]
    ax.plot(angles, values, "o-", linewidth=2, label=region_name, color=colors[idx])
    ax.fill(angles, values, alpha=0.1, color=colors[idx])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=11)
ax.set_title("Average Wage Components by Region (Radar Chart)", fontsize=14, fontweight="bold", pad=30)
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))
ax.set_ylabel("")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(df["density"], df["taxpc"], s=df["crmrte"]*500, c=df["pctmin"],
                      cmap="viridis", alpha=0.6, edgecolors="black", linewidth=0.3)
ax.set_xlabel("Population Density", fontsize=12)
ax.set_ylabel("Tax Per Capita", fontsize=12)
ax.set_title("Bubble Chart: Density vs Tax PC (size=Crime Rate, color=% Minority)", fontsize=13)
cbar = plt.colorbar(scatter)
cbar.set_label("% Minority (pctmin)")
plt.tight_layout()
plt.show()
```
